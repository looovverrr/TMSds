```python
import face_recognition
import os
import pickle

def encode_faces(face_images_dir):
    known_encodings = []
    known_names = []
    for filename in os.listdir(face_images_dir):
        img_path = os.path.join(face_images_dir, filename)
        image = face_recognition.load_image_file(img_path)
        encodings = face_recognition.face_encodings(image)
        if len(encodings) > 0:
            encoding = encodings[0]  
            known_encodings.append(encoding)
            known_names.append("Renee")  
    data = {"encodings": known_encodings, "names": known_names}
    with open("face_encodings.pkl", "wb") as f:
        pickle.dump(data, f)

encode_faces('renee')
```

```python
import cv2
import matplotlib.pyplot as plt

def recognize_faces_in_image(image_path):
    with open("face_encodings.pkl", "rb") as f:
        data = pickle.load(f)
    image = face_recognition.load_image_file(image_path)
    rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    face_locations = face_recognition.face_locations(rgb_image)
    face_encodings = face_recognition.face_encodings(rgb_image, face_locations)
    for (top, right, bottom, left), face_encoding in zip(face_locations, face_encodings):
        matches = face_recognition.compare_faces(data["encodings"], face_encoding)
        name = "Unknown"

        if True in matches:
            first_match_index = matches.index(True)
            name = data["names"][first_match_index]
        cv2.rectangle(rgb_image, (left, top), (right, bottom), (0, 255, 0), 2)
        cv2.putText(rgb_image, name, (left, top - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    plt.imshow(rgb_image)
    plt.axis("off")
    plt.show()
recognize_faces_in_image('timati.png')
```
